## 1.3 Variablen

### Was ist eine Variable?

Eine Variable ist ein **Container**, der einen **Wert** eines bestimmten **Datentyps** aufnehmen kann. Der Zugriff auf die Variable erfolgt über ihren **Bezeichner**.





- **Bezeichner** – der Name der Variable, hier `meineVariable`
- **Zuweisungsoperator** `=` – legt den Wert in der Variable ab
- **Wert** – das, was gespeichert wird, hier `15`

Der Bezeichner steht immer **vor** dem Wert, denn zuerst muss klar sein, *wo* der Wert abgelegt werden soll. Falsch wäre also: ~~`15 = meineVariable`~~

<small>Definition nach Kempe / Löhr (2014: 80) und Schubert / Schwill (2011: 151)</small>

````python
meineVariable = 15
print(meineVariable)
````

### Datentypen

Datentypen werden in Python meist nur **implizit** angegeben: Der Python-Interpreter entscheidet je nach Wert selbst, welchen Datentyp er verwendet. Man kann den Datentyp aber auch **explizit** angeben.

**Vorsicht:** Welchen Datentyp eine Variable hat, kann einen großen Unterschied machen!

| Datentyp | Beschreibung | Datentyp explizit angeben / Typumwandlung |
|:--|:--|:--|
| `int` | Integer – eine Ganzzahl | `x = int(3)` |
| `float` | Dezimalzahl (Gleitkommazahl) – Vorsicht: amerikanische Schreibweise mit Dezimal**punkt** | `x = float(3.1)` |
| `str` | String – eine Zeichenkette (ein Text) | `x = str(3.1)` |
| `bool` | Wahrheitswert (`True` oder `False`) | `x = bool(True)` |

Mit dem Befehl `type()` kann man sich den Datentyp anzeigen lassen:

````python
meineVariable = 15
print(type(meineVariable))
````

### Deklaration und Initialisierung

Bei der **ersten Verwendung** wird eine Variable

- **deklariert** („angemeldet“): Der Bezeichner wird reserviert, ein Datentyp festgelegt (durch Python oder durch dich) und Speicherplatz bereitgestellt.
- **initialisiert**: Sie bekommt ihren ersten Wert – einen sinnvollen Startwert oder direkt eine Nutzereingabe.

In Python geschieht beides in einer einzigen Zeile, wenn man einer Variable das erste Mal einen Wert gibt. Z.B.:

````python
summeEinkaufswagen = 0.0 # Die Variable wird mit 0.0 (€) initialisiert - ein anderer Wert würde zu Beginn kaum sinn machen.

name = input('Name bitte: ')
adresse = ''
bonuspunkte = 0

print('summeEinkaufswagen: ' + str(type(summeEinkaufswagen)))
print('name: ' + str(type(name)))
print('adresse: ' + str(type(adresse)))
print('bonuspunkte: ' + str(type(bonuspunkte)))
````

### Vereinbarung (style convention): Bezeichner für Variablen

- Beginnen mit einem kleinen Buchstaben oder einem Unterstrich
- Enthalten nur alphanumerische Zeichen oder Unterstriche (`a-z`, `A-Z`, `0-9` und `_`)
- Groß- und Kleinschreibung macht einen Unterschied: `varX`, `varx` und `VARX` sind verschiedene Variablen
- Bezeichner sollten „sprechend“ sein, also direkt ihre Verwendung klar machen
- Wörter werden in Variablen durch **Unterstriche** oder „**Camel Case**“ getrennt:

| Unterstriche | Camel Case |
|:--|:--|
| `einkaufswagen_summe` | `einkaufswagenSumme` |
| `hintergrund_farbe` | `hintergrundFarbe` |

**Aufgabe 4** – Das folgende Programm einer Kinokasse ist eigentlich fertig: Es berechnet den Preis für mehrere Tickets, mit Schülerausweis gibt es 50 % Rabatt. Trotzdem bricht es mit einer Fehlermeldung ab.

Der Grund: `input()` liefert immer einen Text (`str`). Die drei Eingaben sollen aber als Dezimalzahl (`float`), als Ganzzahl (`int`) und als Wahrheitswert (`bool`) gespeichert werden.

**a)** Führe das Programm aus und lies die Fehlermeldung. In welcher Zeile tritt der Fehler auf?

**b)** Überlege, welche Eingabe welchen Datentyp braucht, und ergänze in den ersten drei Zeilen die passende Typkonvertierung. Der Rest des Programms ist vollständig und bleibt unverändert.

**c)** Teste dein Programm mit den beiden Eingaben aus der Tabelle. Am Ende gibt das Programm mit `type()` die Datentypen aus – dort müssen `float`, `int` und `bool` stehen.

| Preis pro Ticket | Anzahl | Schülerausweis | erwartete Ausgabe „Zu zahlen“ |
|:--|:--|:--|:--|
| `8.5` | `3` | `1` | 12.75 Euro |
| `8.5` | `3` | `0` | 25.5 Euro |

````python
preis = input('Preis pro Ticket in Euro (z.B. 8.5): ')
anzahl = input('Anzahl der Tickets: ')
hatSchuelerausweis = input('Schülerausweis vorhanden? (1 = ja, 0 = nein): ')

gesamt = float(preis) * int(anzahl)
rabatt = float(gesamt) * 0.5 * bool(hatSchuelerausweis)   # True zählt beim Rechnen als 1, False als 0
endpreis = gesamt - rabatt

print('[Ticket] ' * int(anzahl))
print('Gesamtpreis: ' + str(gesamt) + ' Euro')
print('Rabatt: ' + str(rabatt) + ' Euro')
print('Zu zahlen: ' + str(endpreis) + ' Euro')

print('preis: ' + str(type(preis)))
print('anzahl: ' + str(type(anzahl)))
print('hatSchuelerausweis: ' + str(type(hatSchuelerausweis)))
````

<details>
<summary><b>Tipp 1:</b> Was bedeutet die Fehlermeldung?</summary>

`TypeError: can't multiply sequence by non-int ...` heißt: An der Rechnung ist noch ein Text beteiligt (Python nennt Texte hier „sequence“). Mit Texten kann man so nicht rechnen – die Eingaben müssen erst in Zahlen umgewandelt werden, so wie im Summenprogramm.

Erscheint die Meldung nach deiner ersten Ergänzung erneut, fehlt noch eine weitere Umwandlung.

</details>

<details>
<summary><b>Tipp 2:</b> Der Rabatt wird immer abgezogen – auch bei der Eingabe 0?</summary>

Für Python ist **jeder** nicht-leere Text wahr: `bool('0')` ergibt `True`, genauso wie `bool('nein')` oder `bool('False')`.

Wandle die Eingabe deshalb zuerst in eine Ganzzahl um und diese dann in einen Wahrheitswert: `bool(int(...))`. Aus `1` wird so `True`, aus `0` wird `False`.

</details>

Hinweise:

````python
# Dein Programm zur Berechnung der Tankkosten:
def isinputfloat(message):
    try:
        return float(input(message))       # returns an int
    except ValueError:
        print("Enter a valid number!")   # message, then retry
        return isinputint(message)
liter = isinputfloat("Wieviele Liter wurden getankt?")
preisproliter = isinputfloat("Wieviel kostet es pro Liter")
endpreis = liter * preisproliter
print("Zahlen sie " + str(endpreis) + " € Bitte!!!")
````
