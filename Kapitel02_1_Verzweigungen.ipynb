## 2.1 Einfache Verzweigungen mit Alternative

Eine <b><span style="color:#e00000">Verzweigung</span></b> ist eine Kontrollstruktur, bei der die Ausführung einer Anweisung (oder einer Sequenz aus Anweisungen) an eine <b><span style="color:#158466">Bedingung</span></b> geknüpft ist. Ist sie nicht erfüllt, werden die Anweisungen übersprungen. Es kann auch eine <b><span style="color:#2a6099">Alternative</span></b> angegeben werden, welche Anweisungen stattdessen ausgeführt werden sollen.

Als <b><span style="color:#158466">Bedingung</span></b> nutzt du die **Vergleichsoperatoren** aus Kapitel 1.5 (`<`, `>`, `<=`, `>=`, `==`, `!=`). Ein Vergleich liefert `True` oder `False` – ist das Ergebnis `True`, ist die Bedingung erfüllt.

> **Darauf musst du achten**
>
> - Hinter der Bedingung und hinter `else` steht ein **Doppelpunkt** `:`.
> - Die Anweisungen, die zur Verzweigung gehören, werden **eingerückt** (Tab-Taste oder 4 Leerzeichen). Nur so weiß Python, welche Zeilen dazugehören. `print("Ende")` ist nicht eingerückt und wird deshalb immer ausgeführt.
> - Der `else`-Block kann auch weggelassen werden. Ist die Bedingung dann **nicht** erfüllt, werden keine Anweisungen ausgeführt.
