| `**` | potenz | `2 ** 3` | 8 |
| `+` | positives Vorzeichen | `+1` | +1 |
| `-` | negatives Vorzeichen | `-1` | -1 |

## 1.5 Vergleichsoperatoren

Man kann zwei Zahlen auch mithilfe von **Vergleichsoperatoren** miteinander vergleichen. Heraus kommt dann ein **Wahrheitswert** (`True` oder `False`).

| Operator | Bedeutung | Beispiel: Vergleich | Beispiel: Ergebnis |
|:--|:--|:--|:--|
| `<` | kleiner | `2 < 1` | False |
| `>` | größer | `4 > 2` | True |
| `<=` | kleinergleich | `3 <= 3` | True |
| `>=` | größergleich | `3 >= 4` | False |
| `==` | gleich | `4 == 3` | `False` |
| `!=` | ungleich | `2 != 3` | True |

<details>
<summary><b>Tipp 1:</b> Warum zwei Gleichheitszeichen <code>==</code>?</summary>

Das einfache `=` ist schon vergeben: Es ist der **Zuweisungsoperator** und legt einen Wert in einer Variable ab (`a = 3`).

Das doppelte `==` **vergleicht** dagegen zwei Werte und liefert `True` oder `False`. Eine Verwechslung der beiden ist einer der häufigsten Fehler beim Programmieren!

</details>

<details>
<summary><b>Tipp 2:</b> Wie lese ich <code>!=</code>, <code><=</code> und <code>>=</code>?</summary>

Lies die Zeichen von links nach rechts vor: `!` steht für „nicht“, also `!=` „nicht gleich“. `<=` heißt „kleiner **oder** gleich“.

Gerade bei `<` und `<=` hilft der Fall `a = 3` und `b = 3`: Nur einer der beiden Vergleiche liefert dann `True`.

</details>

### Aufgabe 5 – Notenschnitt

Schreibe ein Python-Programm, mit dem du den Notenschnitt über alle Kernfächer hinweg berechnen kannst. Runde auf eine Nachkommastelle.

Beispiel:

```text
Mathematik: 2
Deutsch: 3
Englisch: 3
Dein Kernfach-Schnitt ist 2.7
```

````python
# Dein Programm zum Notenschnitt:
def isinputfloat(message):
    try:
        return float(input(message))
    except ValueError:
        print("Enter a valid number!")
        return isinputfloat(message)


noten = {}

print("Gib deine Kernfächer und Noten ein (leeres Fach zum Beenden):")

while True:
    fach = input("Fach: ").strip()
    if not fach:
        break
    note = isinputfloat(f"Note für {fach}: ")
    noten[fach] = note

if noten:
    schnitt = round(sum(noten.values()) / len(noten), 1)
    print(f"\nNotenschnitt über alle Kernfächer: {schnitt}")
else:
    print("Keine Fächer eingegeben.")
````

<details>
<summary><b>Tipp 1:</b> Mein Schnitt ist viel zu groß (bei 2, 3, 3 kommt 6.0 heraus).</summary>

Auch Python rechnet **Punkt vor Strich**: Bei `mathe + deutsch + englisch / 3` wird nur die Englischnote durch 3 geteilt.

Setze Klammern, damit zuerst addiert wird: `(mathe + deutsch + englisch) / 3`.

</details>

<details>
<summary><b>Tipp 2:</b> Wie runde ich auf eine Nachkommastelle?</summary>

Mit dem Befehl `round()`. Die Zahl hinter dem Komma gibt an, auf wie viele Nachkommastellen gerundet wird:

```python
round(2.6666, 1)   # ergibt 2.7
```

</details>

### Aufgabe 6 – Body-Mass-Index (BMI)

Der BMI einer Person soll nach der Eingabe des Gewichts in Kilogramm und der Größe in Metern berechnet und ausgegeben werden.

Formel: **BMI = Gewicht / (Größe · Größe)**

Beispiel:

```text
Gewicht in kg: 80.00
Größe in m: 1.80
Der BMI der Person ist 24.69
```

````python
# Dein Programm zum BMI:
def isinputfloat(message):
    try:
        return float(input(message))
    except ValueError:
        print("Enter a valid number!")
        return isinputfloat(message)


gewicht = isinputfloat("Gewicht in kg: ")
groesse = isinputfloat("Größe in m: ")

bmi = gewicht / (groesse * groesse)
print(f"Der BMI der Person ist {bmi:.2f}")
````

<details>
<summary><b>Tipp 1:</b> <code>ValueError: invalid literal for int() with base 10</code></summary>

`1.80` ist keine Ganzzahl. Für Kommazahlen brauchst du `float()` statt `int()`.

Denke außerdem daran: Python schreibt Kommazahlen mit **Punkt**. Wer `1,80` eintippt, bekommt ebenfalls einen Fehler.

</details>

<details>
<summary><b>Tipp 1:</b> Warum ist <code>'5' == 5</code> gefährlich?</summary>

`'5'` ist ein Text, `5` eine Zahl – für Python sind das zwei verschiedene Dinge, also `False`.

Das passiert schnell, wenn du vergisst, eine Eingabe umzuwandeln: Bei `alter = input('Alter: ')` liefert `alter == 18` **immer** `False`, auch wenn jemand 18 eintippt. Python meldet dabei keinen Fehler – das Programm tut einfach still das Falsche.

<details>
<summary><b>Tipp 2:</b> Warum ist <code>0.1 + 0.2 == 0.3</code> falsch?</summary>

Lass dir `0.1 + 0.2` ausgeben: Es kommt `0.30000000000000004` heraus.

Der Computer speichert Dezimalzahlen intern im Zweiersystem. Viele Zahlen wie `0.1` lassen sich dort nicht exakt darstellen – ähnlich wie 1/3 im Zehnersystem (0,333…). Deshalb entstehen winzige Rundungsfehler. Kommazahlen solltest du darum nie mit `==` vergleichen; runde sie vorher, z.B. `round(0.1 + 0.2, 2) == 0.3`.

</details>

### Aufgabe 9 (Zusatz) – Ergebnisse von Vergleichen speichern

Das Ergebnis eines Vergleichs ist ein Wahrheitswert. Den kannst du – wie jeden anderen Wert – in einer Variable speichern. Diese Variable hat dann den Datentyp `bool`.

**a)** Führe das Programm mehrmals mit geraden und ungeraden Zahlen aus. Erkläre, in welcher Reihenfolge Python die zweite Zeile abarbeitet.

````python
zahl = int(input('Gib eine ganze Zahl ein: '))
istGerade = zahl % 2 == 0
print('Die Zahl ist gerade: ' + str(istGerade))
print(type(istGerade))
````

**Deine Antwort** *(Doppelklick zum Bearbeiten, mit Shift + Enter abschließen)*:

zahl % 2 wird zuerst berechnet → der Rest der Division durch 2 (bei geraden Zahlen 0, bei ungeraden 1).

== 0 wird als Nächstes ausgewertet → vergleicht den Rest mit 0. Das ergibt True (Rest ist 0 → gerade) oder False (Rest ist 1 → ungerade).

istGerade = ... → der Wahrheitswert wird ganz zum Schluss der Variable istGerade zugewiesen.

Das liegt an der Operatorrangfolge: % (Modulo) bindet stärker als == (Vergleich). Python rechnet also erst den Rest aus und vergleicht dann – nicht umgekehrt.

**Warum str(istGerade)?**
print kann Strings und Zahlen direkt ausgeben, aber '...' + istGerade würde einen TypeError geben, weil man String und bool nicht mit + verbinden kann. str(...) macht daraus 'True' bzw. 'False' als Text. Alternativ ginge auch print('Die Zahl ist gerade:', istGerade) – dann kümmert sich print selbst um die Umwandlung.
